```python
import dotenv
import os
from openai import OpenAI
```

```python
# In der RENKU-Umgebung muss man nicht diese Zeile laufen lassen
dotenv.load_dotenv()
```

```python
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
```

```python
audio_file = open("../data/gereon.wav", "rb")

transcript_high_german = client.audio.transcriptions.create(
  model="gpt-4o-transcribe",
  file=audio_file
)
```

```python
print(transcript_high_german.text)
```
